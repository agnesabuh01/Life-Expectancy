# How has life expectancy changed over the years?

 - Answer: This dashboard shows how life expectancy increased over the years.

```python
life_by_year = df.groupby('Year')['Life_expectancy'].mean()

plt.figure(figsize=(10, 5))
plt.plot(life_by_year.index, life_by_year.values, marker='o')
plt.title('Average Life Expectancy Over the Years')
plt.xlabel('Year')
plt.ylabel('Average Life Expectancy')
plt.grid(True)
plt.show()
```

## Which regions have the highest life expectancy?
 

```python
region_life = df.groupby('Region')['Life_expectancy'].mean().sort_values(ascending=False)

print(region_life)
```

```python
plt.figure(figsize=(10, 6))
region_life.plot(kind='bar')

plt.title('Average Life Expectancy by Region')
plt.xlabel('Region')
plt.ylabel('Average Life Expectancy')
plt.xticks(rotation=45)
plt.show()
```

## Which 10 countries have the highest life expectancy?

```python
top_10 = (
    df.groupby('Country')['Life_expectancy']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(top_10)
```

```python
plt.figure(figsize=(10, 6))
top_10.sort_values().plot(kind='barh')

plt.title('Top 10 Countries by Average Life Expectancy')
plt.xlabel('Average Life Expectancy')
plt.ylabel('Country')
plt.show()
```

## How does a specific region (e.g., Africa) compare to global averages, and what's changed over time? 



```python
africa = df[df["Region"] == "Africa"].groupby("Year")["Life_expectancy"].mean()

global_avg = df.groupby("Year")["Life_expectancy"].mean()

comparison = pd.DataFrame({
    "Africa": africa,
    "Global": global_avg
})

comparison["Difference"] = comparison["Africa"] - comparison["Global"]

print(comparison)
```

```python
import matplotlib.pyplot as plt

plt.plot(comparison.index, comparison["Africa"], label="Africa")
plt.plot(comparison.index, comparison["Global"], label="Global")

plt.xlabel("Year")
plt.ylabel("Life Expectancy")
plt.title("Africa vs Global Life Expectancy Over Time")

plt.legend()
plt.show()
```

## Which factors (e.g., GDP, schooling, alcohol consumption, BMI, immunization rates) correlate most strongly with life expectancy? 

 - This question is asking you to measure the relationship between each factor and life expectancy and find which ones have the strongest correlation.
 - This dataset was built to examine economic,social,immunization and health factors related to life expectancy. 

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(df["GDP_per_capita"], df["Life_expectancy"])

plt.xlabel("GDP per Capita")
plt.ylabel("Life Expectancy")
plt.title("Life Expectancy vs GDP per Capita")

plt.show()
```

##  Are there notable outlier countries that outperform or underperform relative to their income/schooling level?

 - Which countries have life expectancy much higher or lower than what their GDP per capita and schooling would suggest?

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["Schooling"], df["Life_expectancy"])

plt.xlabel("Schooling")
plt.ylabel("Life Expectancy")
plt.title("Life Expectancy vs Schooling")

plt.show()
```

```python
outliers = df[
    (df["Life_expectancy"] < 50) |
    (df["Life_expectancy"] > 80)
]

print(outliers[["Country", "GDP_per_capita", "Schooling", "Life_expectancy"]])
```

```python
factors = [
    "GDP_per_capita",
    "Schooling",
    "Alcohol_consumption",
    "BMI",
    "Hepatitis_B",
    "Polio",
    "Diphtheria"
]

correlation = df[factors + ["Life_expectancy"]].corr()["Life_expectancy"]

print(correlation.sort_values(ascending=False))
```

```python
country_life = df.groupby("Country")["Life_expectancy"].mean()

top_countries = country_life.sort_values(ascending=False).head(10)

print(top_countries)
```
